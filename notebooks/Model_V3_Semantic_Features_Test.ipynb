```python
import re
from urllib.parse import urlparse


DOC_KEYWORDS = [
    "docs",
    "documentation",
    "tutorial",
    "reference",
    "api",
    "guide",
    "learn"
]


DEV_KEYWORDS = [
    "github",
    "gitlab",
    "npm",
    "pypi",
    "docker",
    "kubernetes",
    "mozilla",
    "python",
    "pallets",
    "tensorflow"
]


SUSPICIOUS_KEYWORDS = [
    "login",
    "signin",
    "verify",
    "verification",
    "secure",
    "account",
    "password",
    "confirm",
    "update"
]


BRAND_KEYWORDS = [
    "paypal",
    "amazon",
    "google",
    "microsoft",
    "apple",
    "facebook"
]


def add_semantic_features(url):

    url_lower = url.lower()

    parsed = urlparse(url)

    path = parsed.path.lower()


    return {

        "HasDocumentationKeyword":
            int(any(
                k in url_lower
                for k in DOC_KEYWORDS
            )),


        "HasDeveloperKeyword":
            int(any(
                k in url_lower
                for k in DEV_KEYWORDS
            )),


        "HasSuspiciousKeyword":
            int(any(
                k in url_lower
                for k in SUSPICIOUS_KEYWORDS
            )),


        "PathDepth":
            len(
                [
                    x for x in path.split("/")
                    if x
                ]
            ),


        "HasBrandKeyword":
            int(any(
                k in url_lower
                for k in BRAND_KEYWORDS
            ))

    }
```

```python
BRAND_DOMAINS = {
    "paypal":"paypal.com",
    "amazon":"amazon.com",
    "google":"google.com",
    "microsoft":"microsoft.com",
    "apple":"apple.com",
    "facebook":"facebook.com"
}


def brand_domain_match(url):

    url_lower=url.lower()

    parsed=urlparse(url)

    domain=parsed.netloc.lower()


    for brand, official in BRAND_DOMAINS.items():

        if brand in url_lower:

            if domain.endswith(official):
                return 1

            else:
                return 0


    return 0
```

```python
import re
from urllib.parse import urlparse


DOC_KEYWORDS = [
    "docs",
    "documentation",
    "tutorial",
    "reference",
    "api",
    "guide",
    "learn"
]


DEV_KEYWORDS = [
    "github",
    "gitlab",
    "npm",
    "pypi",
    "docker",
    "kubernetes",
    "mozilla",
    "python",
    "pallets",
    "tensorflow"
]


SUSPICIOUS_KEYWORDS = [
    "login",
    "signin",
    "verify",
    "verification",
    "secure",
    "account",
    "password",
    "confirm",
    "update"
]


BRAND_DOMAINS = {
    "paypal": "paypal.com",
    "amazon": "amazon.com",
    "google": "google.com",
    "microsoft": "microsoft.com",
    "apple": "apple.com",
    "facebook": "facebook.com"
}
```

```python
def semantic_features(url):

    url_lower = url.lower()

    parsed = urlparse(url_lower)

    domain = parsed.netloc


    path_parts = [
        x for x in parsed.path.split("/")
        if x
    ]


    has_brand = 0
    brand_match = 0


    for brand, official_domain in BRAND_DOMAINS.items():

        if brand in url_lower:

            has_brand = 1

            if domain.endswith(official_domain):
                brand_match = 1

            break



    return {

        "HasDocumentationKeyword":
            int(any(
                k in url_lower
                for k in DOC_KEYWORDS
            )),


        "HasDeveloperKeyword":
            int(any(
                k in url_lower
                for k in DEV_KEYWORDS
            )),


        "HasSuspiciousKeyword":
            int(any(
                k in url_lower
                for k in SUSPICIOUS_KEYWORDS
            )),


        "PathDepth":
            len(path_parts),


        "HasBrandKeyword":
            has_brand,


        "BrandDomainMatch":
            brand_match
    }
```

```python
def extract_features_v3(url):

    old_features = compute_features(
        url,
        tld_lookup,
        char_lookup
    )


    new_features = semantic_features(url)


    return {
        **old_features,
        **new_features
    }
```
